```python
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
df.head()
```

```python
# Question 2: How many records are in the dataset?
num_records = max(df.count())
num_records
```

```python
# Question 3: How many fuel types are presented in the dataset?
fuel_types = set(df['fuel_type'])
fuel_types
```

```python
# Question 4: How many columns in the dataset have missing values?
cols_missing_values = df.columns[num_records - df.count() > 0]
len(cols_missing_values)
```

```python
# Question 5: What's the maximum fuel efficiency of cars from Asia?
max_fuel_efficiency_asia = max(df[df['origin'] == 'Asia']['fuel_efficiency_mpg'])
max_fuel_efficiency_asia
```

```python
# Question 6: Median value of horsepower
"""
1. Find the median value of the horsepower column in the dataset.
2. Next, calculate the most frequent value of the same horsepower column.
3. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
4. Now, calculate the median value of horsepower once again.
"""
horsepower_median = df['horsepower'].median()
horsepower_mode = df['horsepower'].mode().iloc[0]
df['horsepower'] = df['horsepower'].fillna(horsepower_mode)
horsepower_median_new = df['horsepower'].median()
print('median: ', horsepower_median, 'most frequent: ', horsepower_mode, 'median after fillna: ', horsepower_median_new)
```
